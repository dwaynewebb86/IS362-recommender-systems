#Recommender DB

For this assignment, I selected Amazon because its recommendation system is one of the most recognizable examples of a recommender system in e-commerce. Amazon does not only recommend products on the homepage, but across multiple parts of the platform. These include product detail pages, search result pages, cart pages, and email promotions. The platform also uses specific recommendation sections such as “Frequently Bought Together,” “Customers who bought this also bought,” and “Recommended for you.”

Amazon is a strong example because the recommendation system serves two different audiences. The first audience is the customer, who wants to find useful and relevant products quickly. The second audience is Amazon as an organization, which aims to increase engagement, conversions, order size, and customer loyalty.

Performing scenario design twice: once from the customer perspective and once from the organization perspective is beneficial because it allows the analysis to capture both the user experience and the business objectives that drive the recommendation system.

The Amazon recommender system is historically associated with item-to-item collaborative filtering. Instead of finding similar users first, the system identifies similar items based on purchasing, rating, viewing, or other interaction patterns. This approach is important because a large e-commerce platform has millions of customers and products. User-based collaborative filtering can become difficult at scale, since individual users often have sparse and unique purchase histories. Item-to-item collaborative filtering addresses this challenge by focusing on relationships between products and using those relationships to generate recommendations.

A recommender system can also use multiple types of data. These may include purchases, product ratings, browsing behavior, cart activity, and search queries. Additional inputs can include demographic or contextual information, as well as product metadata. Based on this, it is likely that Amazon has expanded beyond the original item-to-item model by incorporating real-time personalization, personalized ranking, complementary product recommendations, and AI-supported shopping features.
